```python
import yt_dlp
import os
import datetime
import shutil

# ===========================
# ⚙️ CONFIGURATION SECTION
# ===========================

# 1. Change this ID for new batches
BATCH_ID = "082" 

# 2. Add specific users you want to sort into their own folders
USERNAMES = ["nja__mbi", "yvonne_njeka", "mnyama.mkali66", "official_nyambura", "dee.sassy1", "eunny_szn", "ivymukash", "queenlavie5",
             "wowcollection01", "trivahnjoki", "lovelylinet1", "lushieynyambu"
             ]

# 3. Define the main location. 
BASE_DIR = os.path.join(os.path.expanduser("~"), "Videos", "tt", BATCH_ID)

# 4. HISTORY LOCATION (Save this right next to the notebook)
# This file will remember every TikTok video ID you download.
NOTEBOOK_DIR = os.getcwd()
HISTORY_FILE = os.path.join(NOTEBOOK_DIR, "tiktok_history.txt")

# ===========================

# Create the main folder if it doesn't exist
os.makedirs(BASE_DIR, exist_ok=True)

# Read the links
link_file = "video_links.txt"
if not os.path.exists(link_file):
    print(f"❌ Error: {link_file} not found!")
    links = []
else:
    with open(link_file, "r", encoding="utf-8") as f:
        links = [line.strip() for line in f if line.strip()]

failed_links = []

print(f"📂 Target Folder: {BASE_DIR}")
print(f"📜 History File:  {HISTORY_FILE}")
print(f"🚀 Starting download for {len(links)} videos...")

# Configure yt-dlp options
ydl_opts = {
    "format": "mp4",
    "noplaylist": True,
    "quiet": False,
    "merge_output_format": "mp4",
    "retries": 3,
    "download_archive": HISTORY_FILE,
    "outtmpl": os.path.join(BASE_DIR, "%(uploader)s_%(id)s.%(ext)s"),
    
    # SWITCHED TO FIREFOX HERE:
    "cookiesfrombrowser": ("firefox",), 
    
    # Adding a modern User-Agent helps "disguise" the script
    "user_agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:120.0) Gecko/20100101 Firefox/120.0"
}

for i, link in enumerate(links, start=1):
    print(f"\n⬇️ [{i}/{len(links)}] Processing: {link}")

    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            # 1. DOWNLOAD & GET INFO
            # If the video is in "tiktok_history.txt", yt-dlp returns None here.
            info = ydl.extract_info(link, download=True)
            
            # 🛑 SKIP CHECK
            if not info:
                print("⏭️ Skipped (Found in History)")
                continue
            
            # 2. LOCATE THE DOWNLOADED FILE
            downloaded_path = ydl.prepare_filename(info)
            
            # Ensure we have the .mp4 extension path
            pre, ext = os.path.splitext(downloaded_path)
            if not os.path.exists(downloaded_path):
                downloaded_path = pre + ".mp4"
            
            if not os.path.exists(downloaded_path):
                raise Exception("Download reported success, but file not found on disk.")

            # 3. SORTING LOGIC
            uploader = info.get("uploader", "unknown")
        
            # We want to remove the '@' if it exists in the name to keep folder names clean
            clean_uploader = uploader.replace("@", "")

            if clean_uploader in USERNAMES:
                # If they are a 'Priority' user from your list, put them in the main folder
                target_subfolder = clean_uploader
            else:
                # If they are anyone else, put them in: others/[username]
                # This satisfies your requirement to group them by username
                target_subfolder = os.path.join("others", clean_uploader)
            
            # Define destination
            destination_folder = os.path.join(BASE_DIR, target_subfolder)
            os.makedirs(destination_folder, exist_ok=True)
            
            filename = os.path.basename(downloaded_path)
            final_path = os.path.join(destination_folder, filename)
            
            # 4. MOVE FILE
            # Check if we are moving it to a different location than where it is
            if os.path.abspath(downloaded_path) != os.path.abspath(final_path):
                # If a file with the same name exists in destination, handle it (optional safety)
                if os.path.exists(final_path):
                    print(f"⚠️ File already exists in {target_subfolder}, overwriting...")
                    os.remove(final_path)
                    
                shutil.move(downloaded_path, final_path)
                print(f"📂 Moved to: \\{target_subfolder}\\")
            else:
                print(f"📂 Kept in: \\{target_subfolder}\\ (Already there)")
                
            print(f"✅ Success")

    except Exception as e:
        print(f"❌ Failed: {link}")
        print(f"   Reason: {e}")
        failed_links.append(link)

# --- LOGGING FAILURES ---
if failed_links:
    ts = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    with open("failed_links.txt", "a", encoding="utf-8") as f:
        f.write("\n")
        f.write(f"{BATCH_ID} [{ts}]\n")
        for l in failed_links:
            f.write(l + "\n")
    print(f"\n⚠️ {len(failed_links)} videos failed. Logged to failed_links.txt")
else:
    print("\n✅ All videos processed!")
```

```python
import ipywidgets as widgets
from IPython.display import display
import webbrowser
import os
import pyperclip

# --- SETUP ---
# We keep the manual path, but the script will now try the system default if this fails
chrome_path = r"C:\Users\HP\AppData\Local\Google\Chrome\Application\chrome.exe"
try:
    webbrowser.register('chrome', None, webbrowser.BackgroundBrowser(chrome_path))
except:
    pass

filename = "failed_links.txt"
# IMPORTANT: This history file should match the one your main script uses!
history_filename = "tiktok_history.txt" 

log_output = widgets.Output()

def load_data_structure():
    """Reads the file and builds a clean list of batch objects."""
    if not os.path.exists(filename):
        return []
    
    with open(filename, "r", encoding="utf-8") as f:
        lines = [line.strip() for line in f if line.strip()]

    batches = []
    current_batch = None
    
    for line in lines:
        # Detect headers like [2025-01-15 10:00:00]
        if "[" in line and "]" in line and not line.startswith("http"):
            current_batch = {"header": line, "links": []}
            batches.append(current_batch)
        elif line.startswith("http") and current_batch is not None:
            current_batch["links"].append(line)
            
    # Clean up empty batches
    batches = [b for b in batches if len(b['links']) > 0]
    return list(reversed(batches))

def delete_link_and_save(target_header, target_link):
    """Removes the specific link from the failed_links.txt file."""
    if not os.path.exists(filename):
        return
    
    # Reload full structure to filter
    all_batches = []
    with open(filename, "r", encoding="utf-8") as f:
        lines = [line.strip() for line in f if line.strip()]
    
    current_batch = None
    for line in lines:
        if "[" in line and "]" in line and not line.startswith("http"):
            current_batch = {"header": line, "links": []}
            all_batches.append(current_batch)
        elif line.startswith("http") and current_batch is not None:
            all_batches.append(line) if current_batch is None else current_batch["links"].append(line)

    # Save back to disk excluding the target link
    with open(filename, "w", encoding="utf-8") as f:
        for b in all_batches:
            if isinstance(b, dict):
                # Only write batch if it still has links after removal
                if b['header'] == target_header:
                    b['links'] = [l for l in b['links'] if l != target_link]
                
                if len(b['links']) > 0:
                    f.write(f"\n{b['header']}\n")
                    for link in b['links']:
                        f.write(f"{link}\n")

def add_to_history(link):
    """Adds the video ID to the history file so yt-dlp skips it next time."""
    try:
        clean_link = link.strip().rstrip('/')
        video_id = clean_link.split('/')[-1]
        
        service = "tiktok" if "tiktok.com" in link else "instagram"
        entry_line = f"{service} {video_id}"
        
        with open(history_filename, "a", encoding="utf-8") as f:
            f.write(f"{entry_line}\n")
        return True, entry_line
    except Exception as e:
        return False, str(e)

def on_video_click(b):
    """Triggered when clicking a link button."""
    link = b.video_link
    header = b.batch_header
    
    # Route to the best downloader
    if "tiktok.com" in link:
        target_url = "https://ssstik.io/en"
    else:
        target_url = "https://fastdl.app/en2"

    with log_output:
        log_output.clear_output()
        
        # 1. COPY LINK
        pyperclip.copy(link)
        print(f"📋 Copied: {link}")

        # 2. OPEN BROWSER (Aggressive Method)
        print(f"🚀 Attempting to open downloader for {'TikTok' if 'tiktok' in link else 'Instagram'}...")
        try:
            # Try Chrome specifically first
            opened = webbrowser.get('chrome').open_new_tab(target_url)
            if not opened: raise Exception("Chrome failed to return True")
        except:
            # Fallback to system default browser (usually more reliable in VS Code)
            webbrowser.open(target_url, new=2)
            
        print(f"💡 Opened: {target_url}")
        print(f"👉 Press Ctrl+V on the page to paste the link.")

        # 3. UPDATE LOGS & FILES
        success, msg = add_to_history(link)
        if success:
            print(f"💾 Saved to History: '{msg}'")
        
        delete_link_and_save(header, link)
        print("🗑️ Removed from failed_links.txt")
        
        # 4. VISUAL BUTTON FEEDBACK
        b.description = "✅ Saved & Removed!"
        b.button_style = 'success'
        b.icon = 'check'
        b.disabled = True

# --- RENDER INTERFACE ---
batches = load_data_structure()

if not batches:
    print("✨ No failed links found. You are all caught up!")
else:
    print(f"📦 Found {len(batches)} batches of failed links.")
    print(f"Instructions: Click a button. It will copy the link and open the downloader site.\n")
    
    container = []
    for batch in batches:
        header = widgets.HTML(f"<b>📂 Batch: {batch['header']}</b>")
        
        btns = []
        for link in batch['links']:
            # Label buttons with the last part of the URL (the ID)
            display_name = link.split('/')[-1] if '/' in link else "Video Link"
            
            btn = widgets.Button(
                description=f"🔗 {display_name}",
                tooltip=link,
                icon='external-link',
                layout=widgets.Layout(width='95%', margin='2px 0px 2px 20px')
            )
            btn.video_link = link
            btn.batch_header = batch['header']
            btn.on_click(on_video_click)
            btns.append(btn)
            
        container.append(widgets.VBox([header] + btns))

    display(widgets.VBox(container))
    print("\nAction Logs:")
    display(log_output)
```

```python
import ipywidgets as widgets
from IPython.display import display
import webbrowser
import os

# --- SETUP CHROME ---
chrome_path = r"C:\Users\HP\AppData\Local\Google\Chrome\Application\chrome.exe"
webbrowser.register('chrome', None, webbrowser.BackgroundBrowser(chrome_path))

filename = "open.txt"
log_output = widgets.Output()

def load_data_structure():
    """Reads open.txt and builds the list."""
    if not os.path.exists(filename):
        return []
    
    with open(filename, "r", encoding="utf-8") as f:
        lines = [line.strip() for line in f if line.strip()]

    batches = []
    current_batch = None
    
    for line in lines:
        # Check for Headers like "Batch [Date]"
        if "[" in line and "]" in line and not line.startswith("http"):
            current_batch = {"header": line, "links": []}
            batches.append(current_batch)
        elif line.startswith("http"):
            # Handle links found before any header
            if current_batch is None:
                current_batch = {"header": "General Links", "links": []}
                batches.append(current_batch)
            current_batch["links"].append(line)
            
    # Filter out empty batches
    batches = [b for b in batches if len(b['links']) > 0]
    return list(reversed(batches))

def on_link_click(b):
    """Runs when you click a button. OPENS ONLY. NO DELETING."""
    link = b.video_link
    
    with log_output:
        log_output.clear_output()
        
        # 1. Open in Chrome
        print(f"🚀 Opening: {link}")
        webbrowser.get('chrome').open(link)
        
        # 2. VISUAL UPDATE ONLY
        # We do NOT delete from file. We do NOT add to history.
        # Just turn green so you know you clicked it.
        b.description = "✅ Opened!"
        b.button_style = 'success' # Green
        b.icon = 'check'
        b.disabled = True

# --- MAIN INTERFACE GENERATOR ---
batches = load_data_structure()

if not batches:
    print(f"✅ '{filename}' is empty or not found.")
else:
    print(f"👇 Click to Open (Links will NOT be deleted):\n")
    
    main_container_items = []
    
    for batch in batches:
        # Header
        header_label = widgets.HTML(f"<b>📂 {batch['header']}</b>")
        
        # Buttons
        link_buttons = []
        for link in batch['links']:
            # Shorten text for cleaner button
            short_text = link.split('/')[-1]
            if not short_text: short_text = "Link"
            
            btn = widgets.Button(
                description=f"🔗 {short_text}",
                tooltip=link,
                icon='external-link',
                layout=widgets.Layout(width='95%', margin='2px 0px 2px 20px') 
            )
            
            btn.video_link = link
            btn.on_click(on_link_click)
            
            link_buttons.append(btn)
        
        # Group them
        batch_group = widgets.VBox([header_label] + link_buttons)
        batch_group.layout.margin = '0px 0px 15px 0px'
        main_container_items.append(batch_group)

    display(widgets.VBox(main_container_items))
    
    print("\nLogs:")
    display(log_output)
```
